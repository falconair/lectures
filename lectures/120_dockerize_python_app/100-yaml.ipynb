### But first, JSON

Programming languages have many useful constructs: loops, if/else conditions, functions. These components are the "machinery" of a programming languages, the way gears are important components of a mechanical system. 

However, we often over look the syntax for data: lists, dictionaries, even basic strings, numbers and booleans:

```python
l = [1, 2, 3, 4, 5]
d = {'george':'washington', 'abe':'lincoln'}
34
"Hi!"
True
```

Configuration files need to contains data but have no need for the "gears" that make a system run - those "gears" belong to full programming languages.

It has occurred to some programmers that only the "data representation" part of programming language might make a great configuration file format!

The first such format was JSON or Javascript Object Notation (pronounced as the name Jason). By coincidence, the example shared above also happens to be correct Javascript code. 

The following is an example from wikipedia:

#### But YAML can do more

However, a good yaml document is supposed to start with `---` 
A few nice syntax changes to improve the data from above:
1. YAML files can have comments
2. The whole document does not need to be wrapped in `{` and `}`
3. You can get away with FAR fewer quotes and commas


```yaml
---

# Client's name
first_name: John
last_name: Smith

# Age related information
is_alive: true
age: 27
```

Additionally, nested dictionaries ("maps" in YAML) can be entered with less syntax, using indentation:

```yaml
...
address:
    street_address: 21 2nd Street
    city: New York
    state: NY
    postal_code: 10021-3100
...
```

YAML also has special syntax for lists:
```yaml
...
children:
    - Catherine
    - Thomas
    - Trevor
...
```


#### Multi-line strings are useful for listing linux commands

My personal favorite feature is YAML's ability to store multiline text. This is useful when YAML is being used to store commands, which naturally spread across multiple lines:

```yaml
...
docker_update_commands: |
 ssh user@remote_host << `EOF`
 docker pull image_name:latest
 docker stop image_name
 docker rm image_name
 docker run -d -restart always -name image_name
EOF
...
```

Note that YAML has multiple variations of multi-line strings

````python
with open('test.yaml') as yaml_file:
    yaml_text = yaml_file.read()
````

````python
yaml_text
````

````python
print(yaml_text)
````

````python
yaml_obj = yaml.safe_load(yaml_text)
yaml_obj
````

````python
docker_commands = yaml_obj['docker_update_commands']
docker_commands
````

````python
print(docker_commands)
````

### Sample YAML file for GitHub Actions

```yaml
name: Build and Push Docker Image

on:
  push:
    branches:
      - main
  pull_request:

jobs:
  build:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout
        uses: actions/checkout@v4

      - name: Extract Docker image metadata
        id: meta
        uses: docker/metadata-action@v5
        with:
          images: ${{ vars.DOCKER_USERNAME }}/my-image

      - name: Log in to Docker Hub
        uses: docker/login-action@v3
        with:
          username: ${{ vars.DOCKER_USERNAME }}
          password: ${{ secrets.DOCKER_PASSWORD }}

      - name: Set up Docker Buildx
        uses: docker/setup-buildx-action@v3
      
      - name: Build and push Docker image
        uses: docker/build-push-action@v6
        with:
          push: ${{ github.event_name != 'pull_request' }}
          tags: ${{ steps.meta.outputs.tags }}
          annotations: ${{ steps.meta.outputs.annotations }}
          provenance: true
          sbom: true
```

````python
with open('github_action_docker.yaml') as gh_yaml_file:
    gh_yaml = yaml.safe_load(gh_yaml_file.read())

gh_yaml
````
