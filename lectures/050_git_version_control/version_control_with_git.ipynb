# Version Control with Git and GitHub

This lecture will teach you how professionals keep track of their code. The tool is called **git**, and the website most people use it with is called **GitHub**.

I should be honest about what this lecture is _not_. We are not trying to turn you into git experts. Git has hundreds of commands and I have used the same dozen for years. These days a coding agent will happily type the commands for you anyway. What the agent will _not_ do is decide for you. It will ask "should I create a branch for this?" or "do you want me to use a worktree?" and you need to know what you are agreeing to.

So the goal is this: when you join a company, you should know what tools you have at your disposal, when to reach for each one, and how to avoid the handful of mistakes that embarrass new hires.

## Why do you need version control?

Have you ever had a folder that looks like this?

<img src="assets/folder_versions.png" width="500"/>

Which one is the latest? Which one did you send to your manager? What is the difference between `version_2_my_data` and `version_2_tommys_data`? Nobody knows, including the person who made them.

Version control solves four problems.

### 1. Version control remembers history
Version control remembers every change you make to your project. You write a short comment explaining _why_ you made each change. You can go back in time to any change. You can even have multiple timelines, where people work on different versions of the code. All of this in a _single_ directory, with no `_version_3_final_FINAL` folders.

### 2. A remote repository is a backup
You can copy the full history of your project to a server. If your laptop falls in a lake, you lose nothing. Your collaborators can copy the project, with its full history, from the same server.

<img src="assets/copy_to_dropbox.png" width="500"/>

Some of you are thinking "I already use Dropbox for this." Dropbox keeps your files safe, but it has no idea what a _change_ is, and it will not help when two people edit the same file.

### 3. A web-based service makes collaboration easier
This is how many of us used to share code:

<img src="assets/share_code_email.png" width="500"/>

Services such as GitHub (also GitLab and BitBucket) replace this. They show you who changed what, let you comment on each other's changes before they are accepted, and give you a place to keep a list of tasks and bugs.

### 4. GitHub is also your resume
Employers often ask for a GitHub link so they can see code you have actually written. Use it to showcase class projects and side projects. But DON'T misrepresent your work! I have personally found copied code in applicants' repositories. It did not go well for them.

## A (very) short history

- **1980s-1990s: RCS, CVS.** Version control existed, but it was clunky and worked one file at a time.
- **2000: Subversion (svn).** One central server holds the history. Your laptop only has the latest copy. This is what I grew up with, and plenty of banks still run it.
- **2005: git.** Linus Torvalds, the creator of Linux, was using a commercial tool called BitKeeper to manage the Linux kernel. The free license was withdrawn, so he wrote his own replacement in a couple of weeks. It had to handle thousands of developers across the world who had never met each other.
- **2008: GitHub.** A website built on top of git. It made git approachable and added the collaboration features we will see later.

### Git is _distributed_
The simple picture of git is hub-and-spoke: there is a server, the "source of truth," and everybody's laptop pushes code to it and pulls code from it. That picture is good enough most of the time.

The more accurate picture is more interesting. When you copy a repository to your laptop, you get the _full history_, not only the latest files. If the server disappears, you have lost nothing. A git system is really "independent repositories talking to other repositories." We will see this with our own eyes in the section _A colleague and a server_, where our "server" will be nothing but a folder on your laptop.

#### Windows users: where are my files?
Git Bash shows you the same folders as File Explorer, but it writes their names the Linux way. Three rules cover almost everything:

| File Explorer says | Git Bash says |
|---|---|
| `C:\Users\shahbaz` (your home folder) | `/c/Users/shahbaz`, or just `~` |
| `C:\Users\shahbaz\Documents\GitHub` | `~/Documents/GitHub` |
| `D:\data` | `/d/data` |

1. The drive letter becomes a folder: `C:\` is `/c/`.
2. Backslashes `\` become forward slashes `/`.
3. `~` is your home folder, `C:\Users\yourname`. This is where Git Bash starts.

So when this notebook says `~/git-lecture`, that is the folder `C:\Users\yourname\git-lecture`, and you can look at it in File Explorer any time you like. Two commands make the connection obvious:

`pwd` tells you where you are, and `explorer .` opens the current folder in File Explorer. (Mac users: `open .` does the same thing in Finder.)

#### **`git config`** tells git who you are
Every change you record is stamped with your name and email. Git will refuse to record anything until it knows them. You only do this once per computer:

```bash
~$ git config --global user.name "Your Name"

~$ git config --global user.email "you@example.com"
```

Use your real name, and the email address you used (or will use) for your GitHub account.

Three more settings, which save you from three common surprises:

```bash
~$ git config --global init.defaultBranch main

~$ git config --global pull.rebase false

~$ git config --global core.editor "code --wait"

~$ git config --global --list
user.name=Your Name
user.email=you@example.com
init.defaultbranch=main
pull.rebase=false
core.editor=code --wait
```

- `init.defaultBranch main`: the main line of your code will be called `main`. **Is it "main" or "master"?** For many years git called it `master`. GitHub switched to `main` in 2020 and most of the industry followed. Git on your laptop may still say `master` unless you set this. You will meet both names in the wild. They mean the same thing.
- `pull.rebase false`: when you and a colleague have both made changes, combine them with a merge. We will see this happen in section 6. Without this setting git stops and asks you to choose.
- `core.editor "code --wait"`: sometimes git needs you to type a message and opens a text editor. This makes that editor VS Code. (Mac users: if `code` is "not found," open VS Code, press `cmd + shift + P`, and run _Shell Command: Install 'code' command in PATH_.)

#### What if I end up in a strange editor I can't leave?
If you skip that last setting, git opens an editor called `vi`. Your screen fills with `~` characters and nothing you type seems to work. Every programmer has been stuck here at least once, usually with someone watching. To get out: press `Esc`, type `:q!` and press enter.

#### SSH keys
Later in this lecture you will send your code to GitHub, and GitHub needs to know it is really you. We do that with an SSH key, the same idea you used to log on to the class server. If you have not yet created a key and added it to your GitHub account, the steps are in the lecture **`060_learn_command_line/generate-ssh-keys.ipynb`**. You do not need it until section 7.

If `git pull` complains instead ("Your local changes would be overwritten"), you have edited the same file yourself. That is not a disaster. Raise your hand, and by the end of section 4 you will know exactly what happened.

### What just happened?
A change went through three stops on my laptop before it could reach yours:

```
        my laptop                                  GitHub              your laptop

  edit a file ---> commit ---------- push ------->  the   ---- pull --->  your
  (step 1)        (step 2)         (step 3)        server                 copy
```

- **Editing** a file changes nothing for anybody else.
- **Committing** saves the change into the history, on my machine only.
- **Pushing** sends my commits to the server.
- **Pulling** brings new commits from the server to you.

That is most of git in one picture. The rest of this lecture is the same journey with you in the driver's seat: sections 2 to 5 happen on your own laptop (edit and commit), and from section 6 on we add a server and other people (push and pull).

### Summary:
  - `git clone address` - copy a repository from a server to your laptop (once)
  - `git pull` - bring the latest changes from the server to your laptop
  - `git log -1` - show the most recent commit
  - A change travels: edit -> commit -> push -> (server) -> pull

### The three places

This is the one idea in git that confuses everybody, so let's go slowly. There are three places a change can be:

```
 working folder              stage                   history
 (what you see in     (what will go into       (commits: saved forever)
  VS Code and ls)      the next commit)

   edit a file  ------ git add ------>   ------ git commit ------>
```

1. **Your working folder** is the ordinary files you see and edit.
2. **The stage** is a waiting area. You put changes here to say "these go into my next save."
3. **The history** is a chain of **commits**. A commit is a saved snapshot of your project with a message explaining it.

Why the middle step? Because you often change five things and only want to save three of them. The stage lets you choose.

#### **`git add`** puts a change on the stage

```bash
~/git-lecture/first$ git add commission.py

~/git-lecture/first$ git status
On branch main

No commits yet

Changes to be committed:    <== on the stage
  (use "git rm --cached <file>..." to unstage)
	new file:   commission.py

Untracked files:    <== not on the stage
  (use "git add <file>..." to include in what will be committed)
	trades.csv
```

`commission.py` is on the stage, `trades.csv` is not.

**Windows users:** you will probably see an extra line here and after other commands, something like `warning: in the working copy of 'commission.py', CRLF will be replaced by LF the next time Git touches it`. Windows and Mac mark the end of a line of text differently, and git is telling you it will store the file in one standard way. It is harmless. Ignore it.

#### **`git commit`** saves what is on the stage into history

```bash
~/git-lecture/first$ git commit -m "Add commission calculator"
[main (root-commit) b36bf7e] Add commission calculator
 1 file changed, 17 insertions(+)
 create mode 100644 commission.py

~/git-lecture/first$ git status
On branch main
Untracked files:
  (use "git add <file>..." to include in what will be committed)
	trades.csv

nothing added to commit but untracked files present (use "git add" to track)
```

The `-m` is for "message." You have made your first commit! Notice that `trades.csv` is _still_ untracked. It was not on the stage, so it was not saved. Let's save it in a second commit:

```bash
~/git-lecture/first$ git add trades.csv

~/git-lecture/first$ git commit -m "Add sample trades"
```

**If your screen fills up and ends with a `:` or the word `(END)`**, git is showing you the output one page at a time. Press `space` for the next page and `q` to quit. This will happen with `git log`, `git diff`, `git show` and `git blame` whenever the output is longer than your window.

Each commit has an author, a date, a message, and a long string of letters and numbers. That string is the commit's **hash**, its unique name. You only ever need the first few characters. Your hashes will be different from mine.

### The most expensive mistake: committing a secret

**Deleting a file does not delete it from history.** That is the whole point of version control! Anyone who can see the repository can see the key, forever.

When this happens with a real key (and it happens every week, to somebody):
1. Tell someone immediately. This is not the time to be shy.
2. The key must be _replaced_ with a new one. Cleaning up the history is not enough, because you cannot know who has already copied it.

In finance the same warning applies to **data**. Vendor data, client positions and trade files usually may not leave the building. They do not belong in a repository.

Look at the log. The mistake is still there, and so is a _new_ commit which reverses it. **Git did not erase history, it added to it.** This matters when you work with others: once your colleagues have a commit, you must never make it disappear. Undo it with a new commit which everybody can see.

Each line shows the commit, the author and the date of its last change. GitHub and VS Code show the same thing with a nicer interface. When you join a team and wonder "why on earth is this here?", this is how you find the commit message which explains it, and the person to ask.

You can visit any old commit. Use the hash of _your_ "Update commission rate" commit, the one with the mistake in it:

  - `git restore filename` - throw away uncommitted changes to a file (no undo!)

Git is protecting you. Switching would overwrite an edit which is saved nowhere. You have three choices:

2. **Throw it away** with `git restore`, if it is not.
3. **Set it aside** with `git stash`, which hides your uncommitted changes in a drawer until you ask for them back with `git stash pop`. You should recognize the word when you see it. I rarely need it.

Ours is not real work:

#### **`git merge name`** brings another branch into the one you are on
We are happy with the new minimum. To bring it into `main`, stand on `main` and pull the other branch in:


```bash

~/git-lecture/branches$ git branch

```

#### "Fast-forward," and why I avoid it
Look at what `git merge` printed: **Fast-forward**. `main` had not changed since we branched off, so git took a shortcut and simply moved `main` forward to the newest commit. Now look at the log. It is a straight line. Nothing in it says that this work was ever done on a branch.

I prefer to keep that information. When I read a history, I want to see "these three commits were the minimum-commission feature" and "these two were the fix for the sell bug." So I merge like this:

```bash
git merge --no-ff higher-minimum
```

`--no-ff` means "no fast-forward." Git then always creates a **merge commit**, which records that a branch came in here, and the graph shows the branch as a visible loop. We will see what that looks like in section 5 (where git has no choice) and we will use `--no-ff` ourselves in section 10.

This is a preference, not a rule. Other people and other teams like the straight line, and git lets you have either. When you join a team, look at how their history looks and do the same. The **Merge pull request** button on GitHub (section 9) creates a merge commit, the way I like it.

### When should you create a branch?
Almost always:
- You are about to start a new feature, a bug fix or an experiment.
- You are about to let a coding agent loose on your code.
- You are not sure whether an idea will work.

Branches cost nothing. Give them names which say what they are for: `higher-minimum`, `fix-sell-commission`, not `test2` or `my-branch`.

Many companies do not even _allow_ you to commit directly to `main`. The server rejects it. The feature-branch habit is how everybody works.

### Summary:
  - `git branch` - list branches (`*` marks the one you are on)
  - `git switch name` - move to a branch
    - `git switch -c name` - create a branch and move to it
  - `git merge name` - bring branch `name` into the branch you are on
    - `git merge --no-ff name` - same, but always record a merge commit (my preference)
  - `git branch -d name` - delete a branch which has been merged
  - `git log --oneline --graph --all` - draw the history of all branches
  - `git stash` / `git stash pop` - set uncommitted changes aside / bring them back
  - `main` is working code. New work starts on a branch

What happens if two branches change the _same line_ in different ways? Git cannot know which one you want, so it stops and asks. This is called a **merge conflict**. It sounds scary. All it means is that a human needs to make a decision.

Conflicts normally happen between two people. To see one calmly, we are going to play both people.

**Person one** works on a branch and lowers the commission to `0.004` because we are moving to a new broker:

```bash
~/git-lecture/conflict$ git switch -c lower-rate
Switched to a new branch 'lower-rate'
```

Edit `commission.py`: change `COMMISSION_PER_SHARE = 0.005` to `COMMISSION_PER_SHARE = 0.004`.

```bash
~/git-lecture/conflict$ git add commission.py

~/git-lecture/conflict$ git commit -m "Lower commission to 0.004 per share"
[lower-rate c304f54] Lower commission to 0.004 per share
 1 file changed, 1 insertion(+), 1 deletion(-)
```

**Person two** is on `main` and knows nothing about that. They renegotiated with the _current_ broker and got `0.0035`:

```bash
~/git-lecture/conflict$ git switch main
Switched to branch 'main'
```

Edit `commission.py`: change `COMMISSION_PER_SHARE = 0.005` to `COMMISSION_PER_SHARE = 0.0035`.

```bash
~/git-lecture/conflict$ git add commission.py

~/git-lecture/conflict$ git commit -m "Lower commission to 0.0035 per share"
[main e47db82] Lower commission to 0.0035 per share
 1 file changed, 1 insertion(+), 1 deletion(-)

~/git-lecture/conflict$ git log --oneline --graph --all
* c304f54 (lower-rate) Lower commission to 0.004 per share
| * e47db82 (HEAD -> main) Lower commission to 0.0035 per share
|/  
* ea9364c Starting point
```

The graph now has a fork: two commits, each changing the same line. Let's merge:

```bash
~/git-lecture/conflict$ git merge lower-rate
Auto-merging commission.py
CONFLICT (content): Merge conflict in commission.py    <== there it is
Automatic merge failed; fix conflicts and then commit the result.

~/git-lecture/conflict$ git status
On branch main
You have unmerged paths.
  (fix conflicts and run "git commit")
  (use "git merge --abort" to abort the merge)

Unmerged paths:
  (use "git add <file>..." to mark resolution)
	both modified:   commission.py

no changes added to commit (use "git add" and/or "git commit -a")
```

Read what `git status` says. It is being quite helpful: we are in the middle of a merge, one file has a conflict, and it lists what we can do about it.

#### **`git merge --abort`** is the way out
Before we fix anything, learn where the exit is. If a merge goes wrong and you want to pretend you never started:

```bash
~/git-lecture/conflict$ git merge --abort

~/git-lecture/conflict$ git status
On branch main
nothing to commit, working tree clean
```

Back to normal, nothing lost. **You can always abort.** Now let's do it properly.

```bash
~/git-lecture/conflict$ git merge lower-rate
Auto-merging commission.py
CONFLICT (content): Merge conflict in commission.py
Automatic merge failed; fix conflicts and then commit the result.

~/git-lecture/conflict$ head -8 commission.py
"""Work out what a day's trades cost us in commission."""
import csv

<<<<<<< HEAD
COMMISSION_PER_SHARE = 0.0035
=======
COMMISSION_PER_SHARE = 0.004
>>>>>>> lower-rate
```

Git has written _both_ versions into the file, separated by markers:

```
<<<<<<< HEAD
COMMISSION_PER_SHARE = 0.0035        <== what the branch you are ON says (main)
=======
COMMISSION_PER_SHARE = 0.004         <== what the branch you are merging IN says
>>>>>>> lower-rate
```

To resolve the conflict you edit the file until it says what it _should_ say, and you remove the three marker lines. You may keep the top version, the bottom version, or write something new.

#### Resolve it in VS Code
Open `commission.py` in VS Code. It highlights the conflict and offers buttons above it: **Accept Current Change** (the top one, `main`), **Accept Incoming Change** (the bottom one, `lower-rate`) and **Accept Both Changes**. This is one of the places where a GUI is simply better than the terminal.

We are staying with our current broker, so click **Accept Current Change**. The line should now read `COMMISSION_PER_SHARE = 0.0035` and the markers should be gone. Save the file.

Then tell git you are done, the same way you record any other change:

```bash
~/git-lecture/conflict$ head -5 commission.py
"""Work out what a day's trades cost us in commission."""

COMMISSION_PER_SHARE = 0.0035
MINIMUM_COMMISSION = 1.00

~/git-lecture/conflict$ python commission.py
Total commission: $51.00

~/git-lecture/conflict$ git add commission.py

~/git-lecture/conflict$ git commit
[main 0e8e2a9] Merge branch 'lower-rate'

~/git-lecture/conflict$ git log --oneline --graph --all
*   0e8e2a9 (HEAD -> main) Merge branch 'lower-rate'
|\  
| * c304f54 (lower-rate) Lower commission to 0.004 per share
* | e47db82 Lower commission to 0.0035 per share
|/  
* ea9364c Starting point
```

(`git commit` without `-m` opens VS Code with a ready-made message, "Merge branch 'lower-rate'". Close the tab to accept it.)

The graph shows the fork closing again. That last commit is a **merge commit**: it has two parents. This is the visible loop I said I like in section 4. Here git had no choice, because both branches had new commits.

Notice that I ran the program _before_ committing. Removing the markers does not mean the code works!

```bash
~/git-lecture/conflict$ git branch -d lower-rate
Deleted branch lower-rate (was c304f54).
```

### Things to remember about conflicts
- A conflict is not an error and nobody did anything wrong. Two people touched the same line.
- Git only asks about lines which truly overlap. Changes to different files, or different parts of one file, merge without any fuss.
- **Never commit a file with `<<<<<<<` in it.** Search for it before you commit. People do this, and it is embarrassing.
- If you are not sure which version is right, _ask the other person_. That is what the conflict is for.
- Lost? `git status`. Really lost? `git merge --abort`.

A sell of 1,000 shares has `shares = -1000`. The commission formula multiplies by the number of shares, gets a _negative_ number, and then `max()` picks the minimum of 1.00 instead. We are charging one dollar on sells which should cost more. Our broker charges on every share traded, bought or sold.

We could just fix it. But in a team, the first step is to **write it down**, somewhere everybody can see it. On GitHub, that place is called an **issue**. (Other companies use Jira or Linear. Same idea, and often the word is "ticket.")

#### Create an issue
On your repository's page on GitHub, click the **Issues** tab, then **New issue**.
- **Title**: `Sells are charged the minimum commission instead of the real one`
- **Description**: say what happens, what should happen, and how to see it. For example:

```
A sell of 1000 shares is charged 1.00 (the minimum). It should be charged 5.00.

Sells have negative shares in trades.csv, so shares * COMMISSION_PER_SHARE is negative
and max() always picks the minimum.

To see it: run python commission.py. It prints 72.00, the correct total is 85.00.
```

Click **Create**. Your issue gets a number: **#1**.

### Why bother?
- **It is a to-do list which lives with the code.** Open issues are what needs doing. Closed issues are a record of what was done, and why.
- **It forces you to say what "done" looks like** before you start. Notice that our issue contains the number we expect once it is fixed.
- **It is the perfect thing to hand to a coding agent.** "Fix issue #1" with a clear issue behind it works far better than a vague chat message, and the agent's work is then linked to the reason for it. Agents can read issues and create them too.
- People who are not programmers (your portfolio manager, the operations team) can read issues, comment on them and create them.

Get in the habit: **no issue, no work**. Even when you work alone.

## 9. Pull requests

(This section continues in `~/git-lecture/github`, with the branch from section 8.)

In section 4 we merged a branch into `main` ourselves, on our own laptop, with nobody looking. In a team, that is not how code gets into `main`. You _ask_.

A **pull request** (everybody says "PR") is a request: "please merge my branch into `main`." It gives your colleagues a page where they can see exactly what you changed, comment on it, ask for improvements, and finally approve it. Git itself has no such thing as a pull request. It is a feature of GitHub. (GitLab calls it a "merge request," which is honestly a better name.)

#### Push your branch to GitHub

```bash
~/git-lecture/github$ git push -u origin fix-sell-commission
...
remote:
remote: Create a pull request for 'fix-sell-commission' on GitHub by visiting:
remote:      https://github.com/YOUR_USERNAME/git-lecture/pull/new/fix-sell-commission
remote:
To github.com:YOUR_USERNAME/git-lecture.git
 * [new branch]      fix-sell-commission -> fix-sell-commission
branch 'fix-sell-commission' set up to track 'origin/fix-sell-commission'.
```

GitHub is practically begging you to open a pull request. `main` on GitHub has not changed. Your branch is sitting next to it.

#### Open the pull request
Go to your repository on GitHub. A yellow banner says your branch "had recent pushes." Click **Compare & pull request**.
- **Title**: GitHub suggests your commit message. Keep it.
- **Description**: what you changed and how you checked it. One or two sentences is fine.
- Check the line at the top: **base: main ← compare: fix-sell-commission**. That reads "merge `fix-sell-commission` into `main`."

Click **Create pull request**.

#### Read it the way a reviewer would
Click the **Files changed** tab. This is `git diff` with a nicer face. Hover over a line and click the blue **+** to leave a comment on that line. Try it: write "Should we add a test for a sell?" and click **Add single comment**.

In a real team, a colleague does this, and you respond. **If they ask for a change, you do not open a new pull request.** You make another commit on the same branch and push it. The pull request updates itself.

#### Merge it
Back on the **Conversation** tab, click **Merge pull request**, then **Confirm merge**. (This button creates a merge commit, the no-fast-forward merge from section 4, so the history will show that the fix arrived through a pull request.) Then click **Delete branch** (the branch on GitHub has done its job).

Now click on the **Issues** tab. Issue #1 is closed, with a link to the pull request which closed it. That is `Fixes #1` at work.

#### Bring your laptop up to date
The merge happened on GitHub. Your laptop does not know yet:

```bash
~/git-lecture/github$ git switch main
Switched to branch 'main'
Your branch is up to date with 'origin/main'.

~/git-lecture/github$ git pull
From github.com:YOUR_USERNAME/git-lecture
   8fdb04b..275463d  main       -> origin/main
Updating 8fdb04b..275463d
Fast-forward

~/git-lecture/github$ git branch -d fix-sell-commission
Deleted branch fix-sell-commission (was 8cb0c0d).

~/git-lecture/github$ git log --oneline --graph

|\  
| * 8cb0c0d (origin/fix-sell-commission) Charge full commission on sells. Fixes #1
|/  
* 8fdb04b Starting point
```

That is the complete loop, and it is how most software teams in the world work:

> **issue → branch → commits → pull request → review → merge → pull**

### How to be good at pull requests
**When you are the author:**
- **Keep them small.** A 50-line pull request gets a careful review. A 2,000-line one gets "looks good to me" from someone who did not read it.
- One pull request, one purpose. Don't mix a bug fix with a reformatting of the whole file.
- Review your own diff before asking anybody else to.
- A review comment is about the code, not about you. Say thanks, fix it, move on.

**When you are the reviewer:**
- Actually read it. You are now partly responsible for this code.
- Ask questions. "Why did you do it this way?" is a perfectly good review comment.
- Be kind and be specific.

#### The green tick and the red cross
On real projects, when you open a pull request, a robot runs the project's tests on your branch. You will see a yellow dot (running), then a green tick (passed) or a red cross (failed) next to your commits. This is called **CI**, for "continuous integration." A red cross means: do not ask anybody to review this yet. We will write the tests which make those ticks later in the course.

#### What if the pull request has a conflict?
If `main` changed in a way which collides with your branch, GitHub shows "This branch has conflicts that must be resolved" and greys out the merge button. The cure is what you already know: on your laptop, `git switch main`, `git pull`, `git switch your-branch`, `git merge main`, resolve as in section 5, and push. The pull request updates itself.

### Summary:
  - A **pull request** asks for your branch to be merged, and gives people a place to review it
  - `git push -u origin branchname` - publish your branch, then open the pull request on GitHub
  - Changes requested? Commit to the same branch and push again
  - After the merge: `git switch main`, `git pull`, `git branch -d branchname`
  - Small pull requests get real reviews
  - The whole loop: issue → branch → commits → pull request → review → merge → pull

## 10. Worktrees: two branches at the same time

> **Logistics:** a fresh copy of the project, just for this section. Nothing to learn here. Copy and paste the whole block into your terminal.

```bash

cp -r starter parallel
cd parallel

git add .

```

In section 4 we learned that there is one folder, and git rewrites its files when you switch branches. That has a consequence: **one folder can only be on one branch at a time.**

Usually that is fine. Here is when it is not:
- A long backtest is running on branch `new-signal`, and your manager needs a one-line fix on `main` _now_. Switch branches and you pull the files out from under the running program.
- **A coding agent is working in your folder.** It is halfway through editing ten files on its branch. You want to carry on with your own work. If you both use the same folder, you will trip over each other. If you start a _second_ agent on another task, it gets worse.

You could clone the repository a second time. That works, but now you have two separate repositories to keep in sync.

A **worktree** is the better answer: **an extra folder, attached to the same repository, checked out on a different branch.** One history, several folders, each on its own branch.

To make the point about ignored files, first give this project a `.gitignore` and a private notes file, as in section 2:

```bash
~/git-lecture/parallel$ echo "notes.txt" > .gitignore

~/git-lecture/parallel$ echo "ask the desk about sells" > notes.txt

~/git-lecture/parallel$ git add .gitignore

~/git-lecture/parallel$ git commit -m "Ignore my scratch notes"
[main 0decdea] Ignore my scratch notes
 1 file changed, 1 insertion(+)
 create mode 100644 .gitignore

~/git-lecture/parallel$ ls
commission.py
notes.txt
trades.csv
```

#### **`git worktree add folder -b branch`** creates a new folder on a new branch
Our project has no README. Let's write one in a worktree, without disturbing the main folder:

```bash
~/git-lecture/parallel$ git worktree add ../parallel-readme -b add-readme
Preparing worktree (new branch 'add-readme')
HEAD is now at 0decdea Ignore my scratch notes

~/git-lecture/parallel$ ls ..
branches
conflict
first
github
github-server.git
parallel
parallel-readme
starter
team
team-colleague
team-server.git
undo

~/git-lecture/parallel$ git worktree list
/Users/you/git-lecture/parallel        0decdea [main]
/Users/you/git-lecture/parallel-readme 0decdea [add-readme]
```

There is a new folder, `parallel-readme`, next to `parallel`. Git lists two worktrees: the original on `main` and the new one on `add-readme`. Go into the new one:

```bash
~/git-lecture/parallel$ cd ../parallel-readme

~/git-lecture/parallel-readme$ ls
commission.py
trades.csv

~/git-lecture/parallel-readme$ git branch
* add-readme
+ main
```

It has the project's files, on its own branch. (The `+` marks a branch which is in use by another worktree.) Look closely at the `ls`: `notes.txt` is missing. Git ignores that file, so git did not bring it along. Remember this, it matters in a moment.

Create a file `README.md` in **this** folder (open it with `code .`, and check that VS Code's title bar says `parallel-readme`):

```

Run `python commission.py` to see what the trades in `trades.csv` cost in commission.
```

```bash
~/git-lecture/parallel-readme$ git add README.md

~/git-lecture/parallel-readme$ git commit -m "Add a README"
[add-readme cc94285] Add a README
 1 file changed, 3 insertions(+)
```

```bash

~/git-lecture/parallel$ ls
```

No `README.md` here. This folder is still on `main` and nothing in it moved, yet the log knows about the new commit. Two folders, **one** repository.

One rule follows from all this. A branch can only be open in one worktree at a time:

```bash
~/git-lecture/parallel$ git switch add-readme
fatal: 'add-readme' is already used by worktree at '/Users/you/git-lecture/parallel-readme'
```

Git refuses, and tells you where that branch is in use.

When the work is finished, it comes back like any other branch (a pull request in real life, a merge here). This is where I use `--no-ff` from section 4:

### "Should I use a worktree?"
Coding agents ask this, in one form or another. Now you know what it means:

> "Should I do this work in a separate folder on its own branch, so that I don't disturb the files you have open?"

Say **yes** when:
- you want to keep working while the agent works,
- you are running more than one agent at once (each gets its own worktree),
- something is running in your main folder which must not be disturbed.

It does not matter when the task is small and you are going to sit and watch it anyway.

Two things to watch:
- **A worktree is a different folder.** Files which git ignores (your `.env`, your `data/` folder, your Python environment) are not in it. A program which needs them will fail there until you copy or re-create them.
- **Clean up.** Finished worktrees are easy to forget. `git worktree list` shows what you have.

### Summary:
  - One folder can only be on one branch at a time. A **worktree** is an extra folder on another branch, sharing the same repository
  - `git worktree add ../folder -b branchname` - create a worktree on a new branch
  - `git worktree list` - show all worktrees
  - `git worktree remove ../folder` - delete a worktree (the branch and its commits stay)
  - A branch can be open in only one worktree at a time
  - Use them to work in parallel: you and an agent, or several agents

## 11. Jupyter notebooks and git

You are reading a Jupyter notebook, and you will write many of them. They need a warning of their own.

A notebook looks like a document, but the `.ipynb` file is really one big blob of text in a format called JSON:

```json
{
  "cells": [
    {
      "cell_type": "markdown",
      "metadata": {},
      "source": [
        "# Let's set up your environment"
      ]
    },
    {
      "cell_type": "code",
      "execution_count": 12,
      "outputs": [ ...every table, every number, every chart, as text... ],
```

This has three consequences:

1. **Diffs are unreadable.** Run a notebook without changing a single word and git sees hundreds of changed lines: execution counts, outputs, images.
2. **Conflicts are miserable.** If two people edit the same notebook, the conflict markers land in the middle of the JSON and the notebook may not even open. Do not share one working notebook between two people. Each person gets their own, or the shared logic goes into a `.py` file.
3. **Outputs are data.** When you display a table of client positions in a notebook and commit it, you have committed client positions. This is the same mistake as the API key in section 2, and much easier to make.

### Habits
- **Clear the outputs before you commit** a notebook: in Jupyter, _Edit → Clear Outputs of All Cells_ (or _Kernel → Restart Kernel and Clear Outputs of All Cells_), then save.
- **Real code belongs in `.py` files.** Use the notebook to explore and to present. When a function is worth keeping, move it to a `.py` file where git can show clean diffs and tests can reach it.
- Add `.ipynb_checkpoints/` to your `.gitignore`.

## 12. Git in the age of coding agents

A coding agent can run every command in this notebook faster than you can type it. So what was the point of the last eleven sections?

The agent does the typing. **You are still the author.** Your name is on the commit, your name is on the pull request, and "the AI wrote it" is not an answer anybody at work will accept. The agent needs you to make the decisions, and you now know what the decisions are.

### What to ask for
- **"Create a branch for this."** Never let an agent work directly on `main`. If it goes badly, you delete the branch.
- **"Work from issue #12."** Write the task down first (section 8). A clear issue is a clear instruction.
- **"Commit as you go, in small commits."** If step four goes wrong, you can go back to step three.
- **"Open a pull request, don't merge it."** Then read it.
- **A worktree**, if you want to keep working meanwhile (section 10).

### What to check
- **Read the diff before anything is committed in your name.** All of it. If you do not understand a change, ask the agent to explain it, or don't accept it.
- **Look at what is being added.** Agents use `git add .` too. Check for data files, keys and stray outputs.
- **Run the code.** A confident summary is not the same as a working program.

### Stop signs
You do not need to know how to use these commands. You need to **recognize** them, and stop and think (or ask someone) before you run them or allow an agent to:

| Command | What it can destroy |
|---|---|
| `git push --force` (or `-f`) | Overwrites the server's history with yours. Your colleagues' commits can vanish |
| `git reset --hard` | Throws away all your uncommitted work, and can throw away commits |
| `git clean -fd` | Deletes every untracked file. There is no recycle bin |
| `git checkout .` / `git restore .` | Throws away all uncommitted changes in every file |
| `git branch -D` | Deletes a branch even though its work exists nowhere else |
| `git rebase` | Rewrites history. Useful in expert hands, not something to try on a shared branch |
| deleting the `.git` folder | Deletes the entire history |

Almost everything else in git can be undone. When an agent asks for permission to run one of these, the right response is "why?"

## The checklist

**Before you start work**
- `git status`: am I where I think I am?
- `git switch main`, `git pull`: start from the latest code
- `git switch -c a-name-that-says-what-this-is`: never work on `main`

**Before you commit**
- `git status`: are these the files I meant to change?
- `git diff`: did I (or my agent) change what I think was changed?
- No passwords, keys, data files or notebook outputs
- No `<<<<<<<` left in any file
- The code runs
- The message says why

**Before you push or ask for a review**
- Pull first. Sort out conflicts on your laptop
- Is the pull request small, with one purpose?
- Have I read my own diff on GitHub?
- Does it mention the issue it fixes?

**When something goes wrong**
- `git status` first. Read what it says
- Mid-merge and confused: `git merge --abort`
- "Detached HEAD": `git switch main`
- Committed something you shouldn't have, and it is already pushed: tell someone, now
- About to type a command you found on the internet containing `--force` or `--hard`: stop, and ask
